```python
import requests
from bs4 import BeautifulSoup
import json
import time
```

```python
url = "https://quotes.toscrape.com/api/quotes"


quotes = [] 
page = 1
while True:

    response = requests.get(url, params={"page": page}, timeout=10)

    response.raise_for_status()
    data = response.json()
    
    quotes.extend(data["quotes"])
    page+=1

    if data["has_next"] == False:
        break

    time.sleep(0.5)

# Load data. This will transform the data into an indented string.
json_data = json.dumps(quotes, indent=2, ensure_ascii=False)

# Create and write to file.
with open("./data/quotes.json", "w", encoding="utf-8") as json_file:
    json_file.write(json_data)
```
